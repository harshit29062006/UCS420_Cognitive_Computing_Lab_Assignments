# 3. Set the random seed to your own roll number using random.seed(your_roll_number) before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

```python
import random as r
r.seed(1024170265)
```

## i. Generate a list of 100 random numbers between 100 and 900 (inclusive).

```python
r.seed(1024170265)
numbers = []
for i in range(100):
    numbers.append(r.randint(100,900))

print(numbers)
```

## ii. Count and print all odd numbers in the list.

```python
odd_numbers = []
for i in numbers:
    if i % 2 != 0:
        odd_numbers.append(i)

print(odd_numbers)
print(len(odd_numbers))
```

## iii. Count and print all even numbers in the list.

```python
even_numbers = []
for i in numbers:
    if i % 2 == 0:
        even_numbers.append(i)

print(even_numbers)
print(len(even_numbers))
```

## iv. Count and print all prime numbers in the list, and additionally build the actual list of prime numbers found using list comprehension.

```python
def is_Prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = []

for i in numbers:
    if is_Prime(i):
        prime_numbers.append(i)

print(prime_numbers)
print(len(prime_numbers))
```

## v. Print the number that occurs most frequently in your list, and how many times it occurs.

```python
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print(most_frequent)
print(frequency)
```

# 4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = {digit × 7 for each digit} and set B = {digit × 9 for each digit}. Print both sets.

```python
C = {1,0,2,4,1,7,0,2,6,5}
A = set()
B = set()

for i in C:
    A.add(i*7)
print(A)

for i in C:
    B.add(i*9)
print(B)
```
